```python
import numpy as np
np.random.seed(42)

# -------- Activation Function --------
def tanh(x):
    return np.tanh(x)

# -------- Inputs --------
# Example input (you can change it if previous network had specific values)
x = np.array([[0.6],
              [0.1]])

# -------- Weights Initialization --------
# Random weights in interval [-0.5, 0.5]

# Input → Hidden (2x2 matrix)
W1 = np.random.uniform(-0.5, 0.5, (2, 2))

# Hidden → Output (1x2 matrix)
W2 = np.random.uniform(-0.5, 0.5, (1, 2))

# -------- Biases --------
b1 = 0.5
b2 = 0.7

# -------- Forward Propagation --------

# Hidden layer
z1 = np.dot(W1, x) + b1
a1 = tanh(z1)

# Output layer
z2 = np.dot(W2, a1) + b2
output = tanh(z2)

# -------- Print Results --------
print("Input:")
print(x)

print(" W1:")
print(W1)

print(" W2:")
print(W2)
print("Network Output:")
print(output)
```
